```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

```python
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Total elements:", temperatures_w1.size)
```

```python
# Convert Celsius to Fahrenheit
fahrenheit = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", fahrenheit)

# Maximum temperature
print("Maximum temperature:", temperatures_w1.max())

# Minimum temperature
print("Minimum temperature:", temperatures_w1.min())

# Mean temperature
print("Mean temperature:", temperatures_w1.mean())
```

```python
#Slicing and indexing
print("First 3 days:", temperatures_w1[:3])
```

```python
print("Weekend:", temperatures_w1[-2:])
```

```python
print("Middle 3 days:", temperatures_w1[2:5])
```

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])

print(temperatures)
```

```python
# Shape
print("Shape:", temperatures.shape)

# Data type
print("Data type:", temperatures.dtype)

# Total number of elements
print("Total elements:", temperatures.size)

# Week 1 temperatures
print("Week 1:", temperatures[0])

# Week 2 temperatures
print("Week 2:", temperatures[1])

# Weekend temperatures for both weeks
print("Weekend temperatures:", temperatures[:, -2:])
```

```python
import pandas as pd

marks = pd.Series([95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])

print(marks)
```

```python
print(marks.iloc[0])
```

```python
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

```python
print(marks.iloc[2])
```

```python
print(marks[marks > 90])
```

```python
##MANIPULATING
marks['Rank1'] = 100
print(marks)
```

```python
marks = marks.drop('Rank5')
print(marks)
```

```python
cgpa = marks / 10
print(cgpa)
```

```python

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Furniture', 'Electronics'],

    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
print(transactions)
print(transactions.head())
print(transactions.tail())
print(transactions.shape)
print(transactions.columns)
print(transactions.dtypes)
```

```python
print(transactions[['ProductCategory', 'Amount']])
```

```python
print(transactions.iloc[:, -3:])
```

```python
print(transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])
```

```python
print(transactions['ProductCategory'].value_counts())
```

```python
print(transactions['Region'].unique())
```

```python
print(transactions.groupby('Region')['Amount'].mean())
```

```python
##MANIPULATION
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions)
```

```python
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)
```

```python
transactions = transactions.drop(transactions[transactions['TransactionID'] == 109].index)
print(transactions)
```

```python
transactions = transactions.drop('Discount', axis=1)
print(transactions)
```
